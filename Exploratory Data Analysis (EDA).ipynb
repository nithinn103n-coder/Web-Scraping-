```python
#Check Missing Values
df.isnull().sum()
```

```python
#Detect Duplicate Rows
df.duplicated().sum()
```

```python
#Industry-wise Company Count
df['Industry'].value_counts()
```

```python
#Visualization
plt.figure(figsize=(10,6))
sns.countplot(y=df['Industry'], order=df['Industry'].value_counts().index)
plt.title("Number of Companies by Industry")
plt.show()
```

```python
#Top 10 Companies by Revenue
top10 = df.sort_values(by='Revenue (USD billions)', ascending=False).head(10)
print(top10[['Name', 'Revenue (USD billions)']])
```

```python
#Visualization
plt.figure(figsize=(12,6))

ax = sns.barplot(
    x='Revenue (USD billions)',
    y='Name',
    data=top10
)

#ax.invert_xaxis()   # Flips graph upside down horizontally
ax.invert_yaxis()
plt.title("Top 10 Companies by Revenue")
plt.show()
```

```python
#Revenue Distribution
plt.figure(figsize=(10,6))
sns.histplot(df['Revenue (USD billions)'], bins=20, kde=True)
plt.title("Revenue Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Revenue (USD billions)'])
plt.title("Revenue Box Plot")
plt.show()
```

```python
#Correlation Analysis
numeric_df = df.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Hypothesis Testing 
sns.scatterplot(x='Employees', y='Revenue (USD billions)', data=df)
plt.title("Employees vs Revenue")
plt.show()
```
